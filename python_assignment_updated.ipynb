## Python’s Role in Data Science
Python plays a central role in data science due to its simplicity, versatility, and powerful ecosystem of libraries. It allows data scientists to efficiently collect, process, analyze, and visualize data.

Key reasons why Python is widely used in data science:
- Easy to learn and use
- Strong libraries like NumPy, Pandas, Matplotlib, Seaborn
- Supports machine learning through libraries like Scikit-learn and TensorFlow
- Large community support

Python enables the complete data science workflow from data cleaning to model deployment.

## Object-Oriented Programming (OOP) Concepts in Python
Object-Oriented Programming (OOP) is a programming paradigm based on the concept of objects and classes.

### Key Concepts:
- **Class**: Blueprint for creating objects
- **Object**: Instance of a class
- **Encapsulation**: Bundling data and methods
- **Inheritance**: Acquiring properties of another class
- **Polymorphism**: Same method behaving differently
- **Abstraction**: Hiding implementation details

OOP helps in writing reusable, modular, and scalable code.

## 6. Replace 'not'...'poor' with 'good'

```python
s='This is not so poor'
not_i=s.find('not')
poor_i=s.find('poor')
if not_i!=-1 and poor_i!=-1 and not_i<poor_i:
    res=s[:not_i]+'good'+s[poor_i+4:]
else:
    res=s
res
```

## 17. Odd & Even series

```python
import math
n=5
odd=sum((2*i)/(math.factorial(2*i-1)) for i in range(1,n+1))
even=sum((2*i)/(math.factorial(2*i)) for i in range(1,n+1))
odd,even
```

## 18. Factorial using recursion

```python
def fact(n):
    return 1 if n==0 else n*fact(n-1)
fact(5)
```
