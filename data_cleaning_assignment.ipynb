## Importance of Data Cleaning in Data Science
Data cleaning is one of the most critical steps in data science. Raw data is often incomplete, inconsistent, and noisy. Without proper cleaning, any analysis or model built on top of the data can produce misleading results.

Data cleaning involves handling missing values, removing duplicates, correcting errors, and ensuring consistency. In real-world scenarios, analysts spend a significant portion of their time cleaning data before actual analysis.

Clean data improves the accuracy of models, ensures reliable insights, and helps in better decision-making. In industries like finance, healthcare, and e-commerce, poor data quality can lead to incorrect predictions and financial loss.

In conclusion, data cleaning is not just a preliminary step but a foundation for all data-driven processes.

## Load Dataset

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/mnt/data/data.csv')
df.head()
```

## Data Overview

```
df.shape, df.info()
```

## Missing Values

```
df.isnull().sum()
```

## Data Cleaning

```
# Drop duplicates
df = df.drop_duplicates()

# Fill numeric missing values
for col in df.select_dtypes(include=np.number).columns:
    df[col].fillna(df[col].median(), inplace=True)

# Fill categorical missing values
for col in df.select_dtypes(include='object').columns:
    df[col].fillna(df[col].mode()[0], inplace=True)

df.isnull().sum()
```

## Exploratory Data Analysis

```
# Distribution of MSRP (assuming column exists)
if 'MSRP' in df.columns:
    sns.histplot(df['MSRP'], kde=True)
    plt.show()
```

```
# Correlation heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```

```
# Example: Top brands by average price
if 'Make' in df.columns and 'MSRP' in df.columns:
    df.groupby('Make')['MSRP'].mean().sort_values(ascending=False).head(10).plot(kind='bar')
    plt.show()
```
